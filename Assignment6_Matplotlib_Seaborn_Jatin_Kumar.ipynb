### Q1. Mathematical Functions Plot
Ask the user to enter a value M.
- Generate x from -10 to 10 using `np.linspace()`.
- Plot y = M*x^2 and y = M*sin(x) in a single figure.

```python
# Ask user to input M
M = float(input('Enter the value of M: '))

# Generate x values from -10 to 10
x = np.linspace(-10, 10, 500)

# Compute y values
y_quadratic = M * x**2       # y = M * x^2
y_sine = M * np.sin(x)       # y = M * sin(x)

# Plot both functions
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y_quadratic, color='steelblue', linestyle='-',  linewidth=2, label=f'y = {M}*x^2')
ax.plot(x, y_sine,      color='tomato',    linestyle='--', linewidth=2, label=f'y = {M}*sin(x)')

ax.set_title(f'Mathematical Functions for M = {M}', fontsize=16, fontweight='bold')
ax.set_xlabel('x', fontsize=13)
ax.set_ylabel('y', fontsize=13)
ax.legend(fontsize=12)
ax.grid(True, linestyle=':', alpha=0.7)
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)

plt.tight_layout()
plt.show()
print(f'Plot displayed for M = {M}')
```

### Q2. Seaborn Bar Plot for Subject Scores
Create a dataset of five subjects and scores, convert to DataFrame,
and plot using Seaborn bar plot with different colors, annotations, title, axis labels, and grid.

```python
# Create dataset of five subjects and scores
data = {
    'Subject': ['Mathematics', 'Physics', 'Chemistry', 'Computer Science', 'English'],
    'Score':   [88, 75, 92, 95, 70]
}

# Convert to Pandas DataFrame
df_subjects = pd.DataFrame(data)
print('Subject Scores DataFrame:')
print(df_subjects)

# Custom colors for each bar
bar_colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B2']

# Seaborn bar plot
fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(data=df_subjects, x='Subject', y='Score',
            hue='Subject', palette=bar_colors, legend=False, ax=ax,
            edgecolor='black', linewidth=0.8)

# Annotate each bar with its score
for bar, score in zip(ax.patches, df_subjects['Score']):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.8,
            str(score),
            ha='center', va='bottom', fontsize=13, fontweight='bold', color='black')

ax.set_title('Student Scores by Subject', fontsize=16, fontweight='bold')
ax.set_xlabel('Subject', fontsize=13)
ax.set_ylabel('Score (out of 100)', fontsize=13)
ax.set_ylim(0, 110)
ax.grid(axis='y', linestyle='--', alpha=0.7)
ax.tick_params(axis='x', labelrotation=15)

plt.tight_layout()
plt.show()
```

### Q3. 2x2 Subplot with Randomized Data
Use roll number as NumPy random seed, generate 50 values using `np.random.randn()`,
and create a 2x2 subplot layout with a line plot (cumulative sum) and scatter plot (with noise).

```python
# Use roll number as seed
ROLL_NUMBER = 1024170438  # Roll number as seed
np.random.seed(ROLL_NUMBER)

# Generate 50 random values
data_random = np.random.randn(50)
print(f'Random seed: {ROLL_NUMBER}')
print(f'First 5 values: {data_random[:5]}')

cumulative_sum = np.cumsum(data_random)
noise = np.random.randn(50) * 0.5
x_idx = np.arange(50)

# Create 2x2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(f'2x2 Subplot | Seed: {ROLL_NUMBER}', fontsize=16, fontweight='bold', y=1.01)

# Subplot 1: Raw data line plot
axes[0, 0].plot(x_idx, data_random, color='steelblue', linewidth=1.8,
                marker='o', markersize=4, label='Random Values')
axes[0, 0].set_title('Random Normal Values', fontsize=13)
axes[0, 0].set_xlabel('Index')
axes[0, 0].set_ylabel('Value')
axes[0, 0].grid(True, linestyle=':', alpha=0.7)
axes[0, 0].legend()

# Subplot 2: Cumulative sum line plot
axes[0, 1].plot(x_idx, cumulative_sum, color='darkorange', linewidth=2, label='Cumulative Sum')
axes[0, 1].fill_between(x_idx, cumulative_sum, alpha=0.2, color='darkorange')
axes[0, 1].set_title('Cumulative Sum (Line Plot)', fontsize=13)
axes[0, 1].set_xlabel('Index')
axes[0, 1].set_ylabel('Cumulative Sum')
axes[0, 1].grid(True, linestyle=':', alpha=0.7)
axes[0, 1].legend()

# Subplot 3: Scatter plot with noise
axes[1, 0].scatter(x_idx, data_random + noise, color='tomato', alpha=0.7,
                   edgecolors='black', linewidths=0.5, label='Data + Noise')
axes[1, 0].set_title('Scatter Plot with Random Noise', fontsize=13)
axes[1, 0].set_xlabel('Index')
axes[1, 0].set_ylabel('Value + Noise')
axes[1, 0].grid(True, linestyle=':', alpha=0.7)
axes[1, 0].legend()

# Subplot 4: Histogram
axes[1, 1].hist(data_random, bins=12, color='mediumseagreen', edgecolor='black', alpha=0.8)
axes[1, 1].set_title('Distribution Histogram', fontsize=13)
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(True, linestyle=':', alpha=0.7, axis='y')

plt.tight_layout()
plt.show()
```

### Q4. Company Sales Data Analysis with Seaborn
Download the dataset from:
https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv

Apply Seaborn to:
1. Line plot of Total Profit per month
2. Multiline plot of all product sales
3. Bar chart for all features/attributes

```python
import urllib.request
import os

# Download dataset
csv_url  = 'https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv'
csv_file = 'company_sales_data.csv'

if not os.path.exists(csv_file):
    urllib.request.urlretrieve(csv_url, csv_file)
    print(f'Dataset downloaded: {csv_file}')
else:
    print(f'Dataset already exists: {csv_file}')

df_sales = pd.read_csv(csv_file)
print('Dataset shape:', df_sales.shape)
print(df_sales.head())
print('Columns:', df_sales.columns.tolist())
```

```python
# Q4.1 - Line plot: Total Profit of all months
fig, ax = plt.subplots(figsize=(12, 5))

sns.lineplot(data=df_sales, x='month_number', y='total_profit',
             marker='o', color='steelblue', linewidth=2.5, markersize=8, ax=ax)

ax.fill_between(df_sales['month_number'], df_sales['total_profit'], alpha=0.15, color='steelblue')
ax.set_title('Total Profit per Month', fontsize=16, fontweight='bold')
ax.set_xlabel('Month Number', fontsize=13)
ax.set_ylabel('Total Profit', fontsize=13)
ax.set_xticks(df_sales['month_number'])
ax.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# Q4.2 - Multiline plot: All product sales data
product_cols = [col for col in df_sales.columns if col not in ['month_number', 'total_profit', 'total_units']]
print('Product columns:', product_cols)

df_melted = df_sales.melt(id_vars='month_number', value_vars=product_cols,
                          var_name='Product', value_name='Sales')

fig, ax = plt.subplots(figsize=(14, 6))

sns.lineplot(data=df_melted, x='month_number', y='Sales', hue='Product',
             marker='o', linewidth=2, markersize=7, ax=ax)

ax.set_title('All Product Sales per Month (Multiline Plot)', fontsize=16, fontweight='bold')
ax.set_xlabel('Month Number', fontsize=13)
ax.set_ylabel('Sales Units', fontsize=13)
ax.set_xticks(df_sales['month_number'])
ax.legend(title='Product', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# Q4.3 - Bar chart for all features/attributes
total_per_product = df_sales[product_cols].sum().reset_index()
total_per_product.columns = ['Product', 'Total Sales']
print('Total sales per product:')
print(total_per_product)

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(data=total_per_product, x='Product', y='Total Sales',
            hue='Product', palette='viridis', legend=False,
            edgecolor='black', linewidth=0.8, ax=ax)

for bar in ax.patches:
    val = int(bar.get_height())
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 200,
            f'{val:,}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')

ax.set_title('Total Sales by Product Category (All Months)', fontsize=16, fontweight='bold')
ax.set_xlabel('Product Category', fontsize=13)
ax.set_ylabel('Total Sales', fontsize=13)
ax.tick_params(axis='x', labelrotation=20)
ax.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
