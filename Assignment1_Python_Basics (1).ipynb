### 3. Arithmetic Operators

```python
a = 15
b = 4

print("a =", a, ", b =", b)
print("Addition (a+b):", a + b)
print("Subtraction (a-b):", a - b)
print("Multiplication (a*b):", a * b)
print("Division (a/b):", a / b)
print("Floor Division (a//b):", a // b)
print("Modulus (a%b):", a % b)
print("Exponentiation (a**b):", a ** b)
```

### 4. String Operations

```python
name = "Jatin Kumar"
print("Original:", name)
print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
print("Length:", len(name))
print("Reversed:", name[::-1])
print("Slice [0:5]:", name[0:5])
print("Replace 'Jatin' with 'JATIN':", name.replace("Jatin", "JATIN"))
```

### 5. Conditional Statements (if-else)

```python
marks = 78

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Marks: {marks} -> Grade: {grade}")
```

### 6. Loops

```python
print("For loop - First 10 natural numbers:")
for i in range(1, 11):
    print(i, end=" ")
print()

print("\nWhile loop - Multiplication table of 5:")
i = 1
while i <= 10:
    print(f"5 x {i} = {5*i}")
    i += 1
```

### 7. Functions

```python
def calculate_simple_interest(principal, rate, time):
    si = (principal * rate * time) / 100
    return si

def is_palindrome(word):
    return word == word[::-1]

print("Simple Interest for P=1000, R=5%, T=2 years:")
print("SI =", calculate_simple_interest(1000, 5, 2))

words = ["madam", "hello", "racecar", "python"]
for w in words:
    print(f"{w} is palindrome: {is_palindrome(w)}")
```

### 8. Lists

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print("Original list:", fruits)
print("Length:", len(fruits))
print("First element:", fruits[0])
print("Last element:", fruits[-1])
print("Sliced [1:3]:", fruits[1:3])

fruits.append("fig")
print("\nAfter append 'fig':", fruits)

fruits.remove("banana")
print("After remove 'banana':", fruits)

fruits.sort()
print("After sorting:", fruits)
```

### 10. Exception Handling

```python
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Cannot divide by zero!"
    except TypeError:
        return "Error: Invalid input type!"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("a", 2))
```
