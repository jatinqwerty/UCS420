### Q1. List Operations
Create a List L = [10, 20, 30, 40, 50, 60, 70, 80] and perform operations.

```python
L = [10, 20, 30, 40, 50, 60, 70, 80]
print("Original List:", L)

L.append(200)
L.append(300)
print("\ni. After adding 200 and 300:", L)

L = [10, 20, 30, 40, 50, 60, 70, 80, 200, 300]  # Reset
L.remove(10)
L.remove(30)
print("ii. After removing 10 and 30:", L)

L_asc = sorted(L)
print("iii. Sorted in ascending order:", L_asc)

L_desc = sorted(L, reverse=True)
print("iv. Sorted in descending order:", L_desc)
```

### Q2. Tuple Operations
Create a tuple scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45) and perform operations.

```python
scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)
print("Scores tuple:", scores)

highest = max(scores)
highest_index = scores.index(highest)
print(f"\ni. Highest score: {highest} at index: {highest_index}")

lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"ii. Lowest score: {lowest}, appears {lowest_count} time(s)")

reversed_list = list(scores[::-1])
print("iii. Reversed tuple as list:", reversed_list)

search_score = 76
if search_score in scores:
    idx = scores.index(search_score)
    print(f"iv. Score {search_score} found at index: {idx}")
else:
    print(f"iv. Score {search_score} is not present in the tuple")
```

### Q3. Random Numbers - Odd, Even, Prime

```python
import random

random_list = [random.randint(100, 900) for _ in range(100)]
print("Generated 100 random numbers between 100 and 900")
print("First 10 numbers:", random_list[:10])

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

odd_numbers = [x for x in random_list if x % 2 != 0]
even_numbers = [x for x in random_list if x % 2 == 0]
prime_numbers = [x for x in random_list if is_prime(x)]

print(f"\ni. All odd numbers ({len(odd_numbers)} total):", odd_numbers)
print(f"\nii. All even numbers ({len(even_numbers)} total):", even_numbers)
print(f"\niii. All prime numbers ({len(prime_numbers)} total):", prime_numbers)
```

### Q4. Set Operations
A = {34, 56, 78, 90}, B = {78, 45, 90, 23}

```python
A = {34, 56, 78, 90}
B = {78, 45, 90, 23}
print("Set A:", A)
print("Set B:", B)

print("\ni. Union (unique scores by both teams):", A.union(B))

print("ii. Intersection (common scores):", A.intersection(B))

print("iii. Symmetric Difference (exclusive scores):", A.symmetric_difference(B))

print("iv. Is A a subset of B?", A.issubset(B))
print("    Is B a superset of A?", B.issuperset(A))

score_to_remove = 56  # example input
if score_to_remove in A:
    A.remove(score_to_remove)
    print(f"v. Removed {score_to_remove} from A. Updated A:", A)
else:
    print(f"v. Score {score_to_remove} is not present in set A.")
```
