## Q1. Sensor readings (no for loops)

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) add the +2 C measurement error to every reading
corrected = temperature + 2
print("a) corrected readings:", corrected)

# b) Celsius -> Fahrenheit, F = 9/5 * C + 32 (on the original readings)
fahrenheit = (9 / 5) * temperature + 32
print("b) in Fahrenheit      :", fahrenheit)

# c) Boolean indexing: readings greater than 32 C
mask = temperature > 32
print("c) mask               :", mask)
print("   readings > 32      :", temperature[mask])

# d) count of readings above 32 (True counts as 1)
print("d) count > 32         :", np.sum(mask), "(same as", mask.sum(), ")")
```

**e) Why vectorization and Boolean indexing are more efficient:** with a `for` loop Python handles one element at a time and runs interpreter code for every step. NumPy applies an operation to the whole array in one call, and the loop runs inside fast compiled C code on data stored in one continuous block of memory. So `temperature + 2` or `temperature > 32` is much faster for big arrays, and the code is also shorter and easier to read. Boolean indexing (`temperature[temperature > 32]`) does the filtering in one line without writing `if` and `append`, and the mask can be reused for counting (`sum`) or for filtering other arrays.

## Q2. Daily steps of four users

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])

print("a) total steps        :", steps.sum())
print("b) mean steps         :", steps.mean())
print("c) max / min          :", steps.max(), "/", steps.min())
print("d) total per day (axis=0)  :", steps.sum(axis=0))
print("e) total per user (axis=1) :", steps.sum(axis=1))

# f) argmax gives the position in the FLATTENED array, so convert it to (row, column)
flat_pos = steps.argmax()
row, col = np.unravel_index(flat_pos, steps.shape)
print("f) argmax (flattened) :", flat_pos)
print(f"   as (user, day)     : ({row}, {col}) -> user {row + 1}, day {col + 1}, value {steps[row, col]}")
```

## Q3. Slicing, views vs copies, reshaping

```python
# a) original array
original = np.array([1, 2, 3, 4, 5, 6])

# b) slice from index 1 to 4 (4 is excluded in Python slicing, so this gives indices 1, 2, 3)
subset = original[1:4]
print("b) subset:", subset)

# c) change first element of subset -> because a slice is a VIEW, original changes too
subset[0] = 999
print("c) original:", original)
print("   subset  :", subset)
```

```python
# d) slice with .copy() -> independent array
original = np.array([1, 2, 3, 4, 5, 6])      # reset so the demo is clean
copied = original[1:4].copy()
copied[0] = 500
print("d) original:", original)
print("   copied  :", copied)
```

```python
# e) numbers 1..12 as a 3 x 4 matrix
m = np.arange(1, 13).reshape(3, 4)
print("e) matrix:\n", m)

# f) indexing and slicing
print("\nf) first row        :", m[0])
print("   last row         :", m[-1])
print("   second column    :", m[:, 1])
# "rows 1-2 and columns 2-3": counting like 'first row / second column' (starting from 1)
print("   rows 1-2, cols 2-3 (counting from 1):\n", m[0:2, 1:3])
print("   same indices counted from 0 (m[1:3, 2:4]):\n", m[1:3, 2:4])
```

```python
# g) flatten() and ravel()
flat = m.flatten()
rav = m.ravel()
print("g) flatten:", flat)
print("   ravel  :", rav)

# h) change an element of the ravel() result -> original matrix changes (ravel returns a view when it can)
rav[0] = 100
print("\nh) after rav[0] = 100, original matrix:\n", m)

# i) change an element of the flatten() result -> original is NOT affected (flatten always returns a copy)
m = np.arange(1, 13).reshape(3, 4)            # fresh matrix
flat = m.flatten()
flat[0] = 100
print("\ni) after flat[0] = 100, original matrix:\n", m)
print("   flat[:4]:", flat[:4])

# j) attributes of the 3 x 4 matrix
print("\nj) shape:", m.shape, "| ndim:", m.ndim, "| size:", m.size, "| dtype:", m.dtype)
```

**Observations:**
- A slice such as `original[1:4]` is a **view**: it shares memory with `original`, so changing `subset[0]` also changed `original[1]` to 999. `.copy()` makes separate memory, so changing the copy left `original` unchanged.
- `ravel()` returns a view here, so changing it **did change** the original matrix. `flatten()` always returns a copy, so changing it **did not** affect the original.

## Q4. Ordinary Least Squares with NumPy

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]])
y = np.array([40, 65, 30, 85])

# a) shape and dimensions
print("a) X.shape:", X.shape, "| X.ndim:", X.ndim, "| 4 users x 3 features")

# b) transpose
print("\nb) X.T =\n", X.T, "\n   shape:", X.T.shape)

# c) X.T @ X
XtX = X.T @ X
print("\nc) X.T @ X =\n", XtX)

# d) inverse (only valid because X.T @ X is square and its determinant is not 0)
print("\nd) det(X.T @ X) =", round(np.linalg.det(XtX), 2))
XtX_inv = np.linalg.inv(XtX)
print("   inverse =\n", XtX_inv)
print("   check (XtX @ inverse = identity):\n", np.round(XtX @ XtX_inv, 6))

# e) OLS: beta = (X^T X)^-1 X^T y
beta = XtX_inv @ X.T @ y
print("\ne) beta (sleep, activity, stress):", beta)
print("   check with np.linalg.lstsq        :", np.linalg.lstsq(X, y, rcond=None)[0])
print("   fitted values X @ beta            :", np.round(X @ beta, 2), "| actual y:", y)
```

```python
# g) predict for the new user
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("g) predicted assistance score for new user:", round(prediction, 2))
```

**b) Transpose:** `X.T` flips rows and columns, so the 4 x 3 matrix (users x features) becomes 3 x 4 (features x users). Each row of `X.T` now holds one feature for all four users. It is needed so that `X.T @ X` gives a 3 x 3 matrix that compares the features with each other.

**d) Inverse:** `np.linalg.inv()` works only for a square matrix with a non-zero determinant. `X` itself is 4 x 3 (not square), so it has no inverse, but `X.T @ X` is 3 x 3 and its determinant is about 753,700, so it can be inverted. (In practice `np.linalg.lstsq` or `np.linalg.solve` is more stable than `inv`, and I used `lstsq` only to check the answer.)

**f) What the coefficients mean:** `beta = [-0.159, 0.141, 10.063]` are the weights of the model `score = b1*sleep + b2*activity + b3*stress`. Each value tells how much the predicted assistance score changes when that feature goes up by one unit and the other two stay the same.
- **Stress (10.06):** by far the strongest effect; one extra stress level adds about 10 points to the assistance score, which fits the data (the users with high stress have the high scores 65 and 85).
- **Sleep (-0.16):** slightly negative, so more sleep gives a slightly lower score, but the size is tiny.
- **Activity (0.14):** slightly positive. This is not very logical for the topic and comes from the data being too small.

Important limits: there are only 4 users for 3 features and the equation has **no intercept term**, so the model is very likely to overfit and the small coefficients should not be trusted too much. Also the features are on different scales (activity is in tens, stress is single digits), so the raw coefficient sizes are not directly comparable.

**g) Prediction:** for `[5, 40, 7]` the predicted assistance score is about **75.28**.
