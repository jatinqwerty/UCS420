### Q2. Basic NumPy Array Operations

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("a) Reversed array:", arr[::-1])

print("\nb) Most Frequent Value:")

x = np.array([1,2,3,4,5,1,2,1,1,1])
print("   i. x =", x)
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_freq = values[counts == max_count]
indices = [np.where(x == v)[0].tolist() for v in most_freq]
print(f"      Most frequent value: {most_freq}, count: {max_count}")
print(f"      Indices: {indices}")

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("\n   ii. y =", y)
values_y, counts_y = np.unique(y, return_counts=True)
max_count_y = counts_y.max()
most_freq_y = values_y[counts_y == max_count_y]
indices_y = [np.where(y == v)[0].tolist() for v in most_freq_y]
print(f"       Most frequent value: {most_freq_y}, count: {max_count_y}")
print(f"       Indices: {indices_y}")
```

### Q4. linspace() Array - Dimensions, Shape, Dtype, Transpose

```python
Jatin = np.linspace(10, 100, 25)
print("Array 'Jatin' (25 evenly spaced numbers from 10 to 100):")
print(Jatin)

print("\nDimensions (ndim):", Jatin.ndim)
print("Shape:", Jatin.shape)
print("Total elements (size):", Jatin.size)
print("Data type (dtype):", Jatin.dtype)
print("Total bytes:", Jatin.nbytes)

print("\nTranspose using reshape():")
transposed_reshape = Jatin.reshape(25, 1)
print(transposed_reshape[:5], "... (shape:", transposed_reshape.shape, ")")

print("\nTranspose using .T attribute (on reshaped 2D):")
arr_2d = Jatin.reshape(5, 5)
print("Reshaped to 5x5:")
print(arr_2d)
print("\nTransposed using .T:")
print(arr_2d.T)
print("\nNote: .T on a 1D array has no effect (shape stays the same)")
```

### Q5. 2D Array - Mean, Median, Max, Min, Unique, Reshape, Resize

```python
ucs420_Jatin = np.array([[10, 20, 30, 40],
                                [50, 60, 70, 80],
                                [90, 15, 20, 35]])

print("ucs420_Jatin (3 rows x 4 columns):")
print(ucs420_Jatin)

print("\nMean:", np.mean(ucs420_Jatin))
print("Median:", np.median(ucs420_Jatin))
print("Max:", np.max(ucs420_Jatin))
print("Min:", np.min(ucs420_Jatin))
print("Unique elements:", np.unique(ucs420_Jatin))

reshaped_ucs420_Jatin = ucs420_Jatin.reshape(4, 3)
print("\nreshaped_ucs420_Jatin (4 rows x 3 columns):")
print(reshaped_ucs420_Jatin)

resized_ucs420_Jatin = np.resize(ucs420_Jatin, (2, 3))
print("\nresized_ucs420_Jatin (2 rows x 3 columns):")
print(resized_ucs420_Jatin)
```
