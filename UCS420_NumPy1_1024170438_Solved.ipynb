## Q2. Basic NumPy array questions

```python
# a) reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])
print("a) reversed:", arr[::-1])
```

```python
# b) most frequent value(s) and their indices
def most_frequent(a):
    values, counts = np.unique(a, return_counts=True)
    top = counts.max()
    for v in values[counts == top]:                 # a loop is used only because there can be a tie
        print(f"   value {v} occurs {top} times at indices {np.where(a == v)[0]}")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("b) i)  x =", x); most_frequent(x)
print("   ii) y =", y); most_frequent(y)
```

For `x`, the value 1 is the most frequent (5 times). For `y`, the values 1 and 3 are tied (3 times each), so both are shown.

## Q4. `linspace`, array attributes and transpose

```python
jatin_kumar = np.linspace(10, 100, 25)       # 25 evenly spaced numbers from 10 to 100
print(jatin_kumar)
print("\nDimensions (ndim)  :", jatin_kumar.ndim)
print("Shape              :", jatin_kumar.shape)
print("Total elements     :", jatin_kumar.size)
print("Data type          :", jatin_kumar.dtype)
print("Total bytes (nbytes):", jatin_kumar.nbytes, "(", jatin_kumar.size, "elements x", jatin_kumar.itemsize, "bytes )")
```

```python
# transpose using reshape(): a 1-D array becomes a column vector
col = jatin_kumar.reshape(25, 1)
print("reshape(25, 1) -> shape:", col.shape)
print(col[:5], "...")

# can we do the same with the T attribute?
print("\njatin_kumar.T shape:", jatin_kumar.T.shape, "-> same as the original, nothing changed")
print("(jatin_kumar.reshape(1, 25)).T shape:", jatin_kumar.reshape(1, 25).T.shape)
```

**Can we do the same with `T`?** Not directly. A 1-D array has only one axis, so `jatin_kumar.T` returns the same array with shape `(25,)`; there is no second axis to swap. `T` works only after the array is 2-D, for example `jatin_kumar.reshape(1, 25).T` gives shape `(25, 1)`. So `reshape()` is the way to turn a 1-D array into a column.

## Q5. 2-D array statistics, reshape and resize

```python
ucs420_jatin_kumar = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(ucs420_jatin_kumar)

print("\nMean   :", ucs420_jatin_kumar.mean())
print("Median :", np.median(ucs420_jatin_kumar))
print("Max    :", ucs420_jatin_kumar.max())
print("Min    :", ucs420_jatin_kumar.min())
print("Unique :", np.unique(ucs420_jatin_kumar))
```

```python
# reshape to 4 rows x 3 columns (the number of elements, 12, stays the same)
reshaped_ucs420_jatin_kumar = ucs420_jatin_kumar.reshape(4, 3)
print("reshaped (4 x 3):\n", reshaped_ucs420_jatin_kumar)

# resize to 2 rows x 3 columns (6 elements, so only the first 6 values are kept)
resized_ucs420_jatin_kumar = np.resize(ucs420_jatin_kumar, (2, 3))
print("\nresized (2 x 3):\n", resized_ucs420_jatin_kumar)
```

**Reshape vs resize:** `reshape` only changes the shape, so the new shape must hold the same number of elements (3 x 4 → 4 x 3 = 12). `resize` can change the number of elements: going to 2 x 3 keeps the first 6 values in row order (10, 20, 30, 40, 50, 60) and drops the rest. If the new size were bigger, `np.resize` would repeat the data to fill it.
